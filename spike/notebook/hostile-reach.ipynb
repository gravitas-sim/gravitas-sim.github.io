```
import js, json
tries = {}
def attempt(name, fn):
    try:
        tries[name] = "allowed: " + str(fn())[:80]
    except Exception as e:
        tries[name] = "refused: " + type(e).__name__ + " " + str(e)[:120]
attempt("parent.document.title", lambda: js.parent.document.title)
attempt("top.localStorage", lambda: js.top.localStorage.length)
attempt("localStorage", lambda: js.localStorage.length)
attempt("document.cookie", lambda: js.document.cookie)
attempt("opener", lambda: js.opener)
attempt("parent.postMessage to *", lambda: js.parent.postMessage("hello from a notebook", "*"))
import pyodide.http
async def f(u):
    r = await pyodide.http.pyfetch(u)
    return r.status
for u in ["http://127.0.0.1:4634/spike/notebook/observation.json", "http://127.0.0.1:4634/instructors/materials.enc.json", "https://example.org/"]:
    try:
        tries["fetch " + u] = "allowed: " + str(await f(u))
    except Exception as e:
        tries["fetch " + u] = "refused: " + type(e).__name__
print(json.dumps(tries, indent=1))
gravitas_output = {"values": tries}
```
